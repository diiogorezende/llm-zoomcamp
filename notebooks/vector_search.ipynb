# Embedding the FAQ dataset

```python
from ingest import load_faq_data
```

```python
documents = load_faq_data()
```

```python
# The data is on dict format
documents[10]
```

```python
# Combine the question and answer fields from each document into a single text string
texts = []

for doc in documents:
    text = doc["question"] + " " + doc["answer"]
    texts.append(text)
```

```python
texts
```

```python
# Turn the combined texts into embeddings
from tqdm.auto import tqdm

batch_size = 50
vectors = []

for i in tqdm(range(0, len(texts), batch_size)):
    batch = texts[i : i + batch_size]
    batch_vectors = model.encode(batch)
    vectors.extend(batch_vectors)
len(vectors)
```

```python
vectors[10].shape
```

```python
# Convert the list of vectors to a NumPy array for efficient computation
X = np.array(vectors)
```

```python
X
```

```python
# Compute the similarity scores between the query embedding and all vectors in the dataset
scores = X.dot(q1_embedding)
```

```python
scores
```

```python
# Selecting the most similar vector based on the computed similarity scores
most_similar_index = np.argmax(scores)
most_similar_index, scores[most_similar_index], texts[most_similar_index]
```

```python
documents[1004]
```

```python
# Getting the top 5 most similar vectors
top_5_indices = np.argsort(scores)[-5:][::-1]
top_5_indices, scores[top_5_indices], [texts[i] for i in top_5_indices]
```

__Vector Search with FAISS__

```python
import faiss
```

```python
X.shape
```

```python
dim = X.shape[1]
index = faiss.IndexFlatIP(dim)  # Dot product similarity
index.add(X.astype(np.float32))
```

```python
query_vector = q1_embedding.astype(np.float32).reshape(1, -1)
scores, indices = index.search(query_vector, k=5)
```

```python
top_5_docs = [documents[i] for i in indices[0]]
```

```python
top_5_docs
```

```python
# Adding filtering for a specific course
course_ids = np.array(
    [i for i, doc in enumerate(documents) if doc["course"] == "llm-zoomcamp"],
    dtype=np.int32,
)
```

```python
id_selector = faiss.IDSelectorArray(course_ids)
params = faiss.SearchParameters(sel=id_selector)
```

```python
scores, indices = index.search(query_vector, k=5, params=params)
top_5_docs = [documents[i] for i in indices[0]]
```

```python
top_5_docs
```

__RAG with Vector Search__

```python
from rag_helper import RAGHelper
from ingest import load_faq_data, build_faiss_index
from dotenv import load_dotenv
from anthropic import Anthropic
import os
```

```python
load_dotenv()
```

```python
anthropic_client = Anthropic(api_key=os.getenv("ANTHROPIC_API_KEY"))
```

```python
documents = load_faq_data()
index_faiss = build_faiss_index(model=model, documents=documents)
```

```python
class RAGVectorSearch(RAGHelper):
    def __init__(self, embedder, documents, **kwargs):
        super().__init__(**kwargs)
        self.embedder = embedder
        self.documents = documents

    def search(self, query, num_results=5):
        query_vector = self.embedder.encode(query).astype(np.float32).reshape(1, -1)
        course_ids = np.array(
            [i for i, doc in enumerate(self.documents) if doc["course"] == self.course],
            dtype=np.int32,
        )
        id_selector = faiss.IDSelectorArray(course_ids)
        params = faiss.SearchParameters(sel=id_selector)

        _, indices = self.index.search(query_vector, num_results, params=params)
        return [self.documents[i] for i in indices[0] if i != -1]
```

```python
vector_assistant = RAGVectorSearch(
    embedder=model, index=index_faiss, documents=documents, llm_client=anthropic_client
)
```

```python
query = "I just found out about the course, can I still sign up?"
```

```python
vector_assistant.search(query)
```

__Vector Search with sqlitesearch__

```python
import sqlite3
import sqlite_vec
```

```python
db = sqlite3.connect("vector_search.db")
db.enable_load_extension(True)
sqlite_vec.load(db)
db.enable_load_extension(False)
```

```python
db.execute("""
CREATE VIRTUAL TABLE IF NOT EXISTS faq_vectors USING vec0(
    embedding FLOAT(384),
    course TEXT
)
""")
```

```python
for doc, vec in zip(documents, X):
    db.execute(
        "INSERT INTO faq_vectors (embedding, course) VALUES (?, ?)",
        (vec.astype(np.float32).tobytes(), doc["course"]),
    )
```

```python
db.commit()
```

```python
query_vector = q1_embedding.astype(np.float32).tobytes()

rows = db.execute(
    """
    SELECT rowid, distance
    FROM faq_vectors
    WHERE embedding MATCH ? AND course = ? AND k = 5
    ORDER BY distance
""",
    (query_vector, "llm-zoomcamp"),
).fetchall()
```

```python
top_5_docs = [documents[rowid - 1] for rowid, _ in rows]
```
