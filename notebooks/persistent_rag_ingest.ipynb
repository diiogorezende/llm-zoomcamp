```python
from ingest import load_faq_data
import time
from sqlitesearch import TextSearchIndex
```

```python
documents = load_faq_data()
```

```python
docs_llm = [doc for doc in documents if doc["course"] == "llm-zoomcamp"]
print(f"Loaded {len(docs_llm)} documents for llm-zoomcamp course.")
```

```python
index = TextSearchIndex(
    text_fields=["question", "section", "answer"],
    keyword_fields=["course"],
    db_path="faq.db",
)
```

```python
for doc in docs_llm:
    index.add(doc)
    print(f"Added: {doc['question'][:60]}...")
    time.sleep(0.5)  # Sleep for 0.5 seconds to avoid overwhelming the database
```

```python
index.close()  # Close the index to ensure all data is written to the database
print("Ingestion complete. All documents have been added to the database.")
```
