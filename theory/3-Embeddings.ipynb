# __Embeddings__
In NLP, a token is the basic unit of a language model. A token can be a character, a word, a part of word, depending on the model. Tokens are classicalyy represented as sparce vectors and has a vocabulary size that can easily grow into the millions.

An embedding is the process of turning text, images, or data into a list of numbers (a vector) in a way that the computer can understand the meaning of it. Embedding are learned continuous vector representations of discrete data. This means that instead of treating those entities (like words, categorical features, images, etc) as sparse, high-dimensional, and mutually orthogonal encoded vectors, an embedding maps them into a dense, lower-dimensional vector where the geometric distance directly correlates with semantic similarity.

The embeddings are not engineered by hand, they are learned. There are many ways to learn those embeddings, but I will bring one:
- __Contextual Encoding:__ This is the _Transformer_. Rather than lookups, like BERT models, Transformers uses a self-attention mechanism to generate dinamic embeddings. With this approach, the final vector of a given word depend on whether the surrounding tokens relate to it.

Once trained, those vectors enable what we call __Semantic Search__. Documents or objects are embedded and stored in a vector index. When a user issues a query, it is mapped to the same space, the system performs an Approximate Nearest Neighbor (ANN) search to retrieve the closest vectors.
